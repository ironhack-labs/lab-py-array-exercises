1. **Find Maximum and Minimum in an Array**
    - Write a Python program to find the maximum and minimum values in an integer array.

    **Example:**
    ```python
    # Sample Input
    arr = [10, -2, -30 ,4 ,100 ,50]

    # Sample Output
    Maximum value:100
    Minimum value:-30
    ```


````python
# Write your code HERE
integer_array = arr.array('i', [10, -2, -30 ,4 ,100 ,50])

def find_maximum(arr):
    if len(arr) == 0:
        return None
    max_value = arr[0]
    for num in arr:
        if num > max_value:
            max_value = num
    return max_value
maximum_value = find_maximum(integer_array)
print("Maximum value in the array:", maximum_value)
````

````python
def find_minimum(arr):
    if len(arr) == 0:
        return None
    min_value = arr[0]
    for num in arr:
        if num < min_value:
            min_value = num
    return min_value
minimum_value = find_minimum(integer_array)
print("Minimum value in the array:", minimum_value)
````


2. **Count Even and Odd Numbers in an Array**
    - Write a Python program to count how many even and odd numbers are present in an integer array.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,21 ,32 ,43 ,54 ,65 ,76 ,87]

    # Sample Output
    Even count:4
    Odd count:4
    ```


````python
# Write your code HERE
integer_array = arr.array('i', [10 ,21 ,32 ,43 ,54 ,65 ,76 ,87])

def count_even_numbers(arr):
    count = 0
    for num in arr:
        if num % 2 == 0:
            count += 1
    return count
even_count = count_even_numbers(integer_array)
print("Count of even numbers in the array:", even_count)

def count_odd_numbers(arr):
    count = 0
    for num in arr:
        if num % 2 != 0:
            count += 1
    return count
odd_count = count_odd_numbers(integer_array)
print("Count of odd numbers in the array:", odd_count)
````

1. **Rotate Array Elements**
    - Write a Python program to rotate elements of an integer array by `k` positions.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,20 ,30 ,40 ,50]
    k =2

    # Sample Output
    Rotated Array: [30 ,40 ,50 ,10 ,20]
    ```


````python
# Write your code HERE
integer_array = arr.array('i', [10 ,20 ,30 ,40 ,50])
k = 2
def rotate_array(arr, k):
    n = len(arr)
    k = k % n  # Handle cases where k is greater than n
    return arr[-k:] + arr[:-k]
rotated_array = rotate_array(integer_array, k)
print("Rotated array:", rotated_array)
````


2. **Merge Two Sorted Arrays**
     - Write a Python program to merge two sorted arrays into one sorted array.

     **Example:**
     ```python
     # Sample Input
     arr1 = [1 ,3 ,5]
     arr2 = [2 ,4 ,6]

     # Sample Output
     Merged Sorted Array: [1 ,2 ,3 ,4 ,5 ,6]
     ```



````python
# Write your code HERE
array1 = arr.array('i', [1, 3, 5])
array2 = arr.array('i', [2, 4, 6])
def merge_arrays(arr1, arr2):
    return sorted(arr1 + arr2)
merged_array = merge_arrays(array1, array2)
print("Merged array:", merged_array)
````
