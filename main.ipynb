
8. **Check for Duplicates in Array**
    - Write a Python program to find out if a given array of integers contains any duplicate elements.

    **Example:**
    ```python
    # Sample Input
    arr = [1,2,3,4]

    # Sample Output
    Contains duplicates: False

    arr = [1,2,2]

    Contains duplicates: True
    ```


````python
# Write your code HERE

def find_duplicates(arr):
    for el in arr:
        count = 0
        for k in arr:
            if k == el:
                count += 1
        
        if count > 1:
            return True
    return False
    
find_duplicates([1,2,3,4,4])
````

## Advanced Exercises



1. **Find Maximum and Minimum in an Array**
    - Write a Python program to find the maximum and minimum values in an integer array.

    **Example:**
    ```python
    # Sample Input
    arr = [10, -2, -30 ,4 ,100 ,50]

    # Sample Output
    Maximum value:100
    Minimum value:-30
    ```


````python
# Write your code HERE
arr = [10, -2, -30 ,4 ,100 ,50]

def find_min(arry):
    min = arry[0]
    for num in arry:
        if num < min:
            min = num
    return min

def find_max(arry):
    max = arry[0]
    for num in arry:
        if num > max:
            max = num
    return max

print("Maximum value:", find_max(arr))
print("Minimum value:", find_min(arr))
````


2. **Count Even and Odd Numbers in an Array**
    - Write a Python program to count how many even and odd numbers are present in an integer array.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,21 ,32 ,43 ,54 ,65 ,76 ,87]

    # Sample Output
    Even count:4
    Odd count:4
    ```


````python
# Write your code HERE
arr = [10 ,21 ,32 ,43 ,54 ,65 ,76 ,87, 88]

def count_even (arr):
    count = 0
    for el in arr:
        if el % 2 == 0:
            count += 1
    
    return count

def count_uneven(arr):
    count = 0
    for el in arr:
        if el % 2 != 0:
            count += 1
    
    return count


print("Even count:", count_even(arr))
print("Odd count:", count_uneven(arr))
````

1. **Rotate Array Elements**
    - Write a Python program to rotate elements of an integer array by `k` positions.

    **Example:**
    ```python
    # Sample Input
    arr = [10 ,20 ,30 ,40 ,50]
    k =2

    # Sample Output
    Rotated Array: [30 ,40 ,50 ,10 ,20]
    ```


````python
# Write your code HERE

def roate_array(arr, k):
    new_array = []
    for i in range(k, len(arr)):
        new_array.append(arr[i])
    
    for i in range (0, k):
        new_array.append(arr[i])
    
    return new_array


arr = [10 ,20 ,30 ,40 ,50]
roate_array(arr, 2)
````


2. **Merge Two Sorted Arrays**
     - Write a Python program to merge two sorted arrays into one sorted array.

     **Example:**
     ```python
     # Sample Input
     arr1 = [1 ,3 ,5]
     arr2 = [2 ,4 ,6]

     # Sample Output
     Merged Sorted Array: [1 ,2 ,3 ,4 ,5 ,6]
     ```



````python
# Write your code HERE
arr1 = [1 ,3 ,5]
arr2 = [2 ,4 ,6]

def merge_and_sort(arr1, arr2) -> list[int]:
    new_array = []
    for el in arr1:
        new_array.append(el)        
    for el in arr2:
        new_array.append(el)    
    sorted_array = sorted(new_array)
    return sorted_array

print("Merged Sorted Array:" , merge_and_sort(arr1, arr2 ))
````
